```python

!pip install -q spacy nltk vaderSentiment
!python -m spacy download en_core_web_sm

import spacy
from nltk.stem import PorterStemmer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer
from IPython.display import display, HTML
import ipywidgets as widgets

nlp = spacy.load("en_core_web_sm")
stemmer = PorterStemmer()
analyzer = SentimentIntensityAnalyzer()

print("NLP Chatbot is ready!")

def analyze_sentence(sentence):

    if not sentence.strip():
        print("Please enter a sentence.")
        return

    doc = nlp(sentence)

    print("\n========== NLP CHATBOT ==========")

    # Tokenization
    print("\n1. TOKENIZATION")
    print([t.text for t in doc])

    # Stemming
    print("\n2. STEMMING")
    print([stemmer.stem(t.text) for t in doc if t.is_alpha])

    # Lemmatization
    print("\n3. LEMMATIZATION")
    print([t.lemma_ for t in doc if t.is_alpha])

    # NER
    print("\n4. NAMED ENTITY RECOGNITION")
    if doc.ents:
        for ent in doc.ents:
            print(ent.text, ":", ent.label_)
    else:
        print("No entities found.")

    # Sentiment
    print("\n5. SENTIMENT ANALYSIS")
    scores = analyzer.polarity_scores(sentence)

    if scores["compound"] >= 0.05:
        print("Positive 😊")
    elif scores["compound"] <= -0.05:
        print("Negative 😞")
    else:
        print("Neutral 😐")

    print("Sentiment scores:", scores)
    print("\n=================================")


sentence_input = widgets.Textarea(
    placeholder="Enter your sentence here...",
    description="Sentence:",
    layout=widgets.Layout(width="90%", height="100px")
)

analyze_button = widgets.Button(
    description="Analyze Sentence",
    button_style="success",
    icon="search"
)

output = widgets.Output()

def on_click(button):
    with output:
        output.clear_output()
        analyze_sentence(sentence_input.value)

analyze_button.on_click(on_click)

display(sentence_input, analyze_button, output)
```
